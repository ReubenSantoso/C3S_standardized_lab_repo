This noteboook is a sample on how to use the c3sdb module to standardize training, validation, and test code.

```python
from c3sdb import C3SDB
```

```python
# Then use it like:
model = C3SDB("C3S_clean_smile.db")
```

```python
model.build_features(feature_columns=["adduct", "mz"], target_column="ccs", use_mqns=True)

#checking feature building
print(model.X_columns)
print(model.y_column)
print(model.db.columns)
```

```python
model.train_test_split(stratify="source") #stratisfy on source
# model.train_test_split() #stratisfy on ccs
```

```python
# Shapes of datasets with clarifying print statements
print("Shape Of Total Dataset (features):", model.X_df.shape)
print("Shape Of Total Dataset (targets):", model.y_df.shape)

print("Shape Of Training Set (features):", model.trainX_df.shape)
print("Shape Of Training Set (targets):", model.trainY_df.shape)

print("Shape Of Test Set (features):", model.testX_df.shape)
print("Shape Of Test Set (targets):", model.testY_df.shape)

print("Shape Of Validation Set (features):", model.validationX_df.shape)
print("Shape Of Validation Set (targets):", model.validationY_df.shape)
```

```python
model.center_and_scale()
```

```python
model.train(model.trainX_df, model.trainY_df) #implement the exact method on every project
```

```python
# Computer prediction metrics on validation set
pred = model.predict(model.validationX_df, model.validationY_df)

results = model.compute_metrics(y_pred=pred, y_true=model.validationY_df, dataset="validation")
print(results)

model.save_predictions_to_csv("validation")
```

```python
#Compute prediction metrics on test set
pred = model.predict(model.testX_df, model.testY_df)

results = model.compute_metrics(y_pred=pred, y_true=model.testY_df, dataset="test")
print(results)

model.save_predictions_to_csv("test")
```
